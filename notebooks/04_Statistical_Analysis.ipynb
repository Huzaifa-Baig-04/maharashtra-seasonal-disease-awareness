```python
# ============================================================
# CEP PROJECT
# Predictive Analysis of Seasonal Diseases in Maharashtra
#
# FILE 4: STATISTICAL ANALYSIS
# ============================================================

# ------------------------------------------------------------
# 1. MOUNT GOOGLE DRIVE
# ------------------------------------------------------------

from google.colab import drive
drive.mount('/content/drive')


# ------------------------------------------------------------
# 2. IMPORT LIBRARIES
# ------------------------------------------------------------

import os
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import spearmanr, f_oneway

warnings.filterwarnings("ignore")


# ------------------------------------------------------------
# 3. PROJECT PATHS
# ------------------------------------------------------------

PROJECT_PATH = "/content/drive/MyDrive/CEP Seasonal Disease Prediction"

PROCESSED_PATH = os.path.join(
    PROJECT_PATH,
    "Data",
    "processed"
)

INPUT_FILE = os.path.join(
    PROCESSED_PATH,
    "Maharashtra_Seasonal_Disease_Cleaned.csv"
)

STAT_PATH = os.path.join(
    PROCESSED_PATH,
    "Statistical_Analysis"
)

os.makedirs(STAT_PATH, exist_ok=True)

print("Project Path:")
print(PROJECT_PATH)

print("\nInput File:")
print(INPUT_FILE)

print("\nOutput Folder:")
print(STAT_PATH)


# ------------------------------------------------------------
# 4. LOAD CLEANED DATASET
# ------------------------------------------------------------

df = pd.read_csv(INPUT_FILE)

print("\nDataset loaded successfully.")

print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())


# ------------------------------------------------------------
# 5. BASIC DATA PREPARATION
# ------------------------------------------------------------

# Convert Date if available
if "Date" in df.columns:
    df["Date"] = pd.to_datetime(
        df["Date"],
        errors="coerce"
    )

# Convert numerical columns
numeric_columns = [
    "Cases",
    "Deaths",
    "preci",
    "LAI",
    "Temp",
    "week_of_outbreak"
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )

print("\nData types after conversion:")
print(df[numeric_columns].dtypes)


# ------------------------------------------------------------
# 6. CREATE BASIC STATISTICAL SUMMARY
# ------------------------------------------------------------

available_numeric = [
    col for col in numeric_columns
    if col in df.columns
]

statistical_summary = df[available_numeric].describe().T

statistical_summary["missing_values"] = (
    df[available_numeric].isna().sum()
)

statistical_summary["missing_percentage"] = (
    df[available_numeric].isna().mean() * 100
)

statistical_summary = statistical_summary.round(4)

print("\nStatistical Summary:")
display(statistical_summary)

statistical_summary.to_csv(
    os.path.join(
        STAT_PATH,
        "statistical_summary.csv"
    )
)


# ------------------------------------------------------------
# 7. PEARSON CORRELATION MATRIX
# ------------------------------------------------------------

correlation_data = [
    col for col in [
        "Cases",
        "Deaths",
        "preci",
        "LAI",
        "Temp",
        "week_of_outbreak"
    ]
    if col in df.columns
]

correlation_matrix = df[
    correlation_data
].corr(method="pearson")

print("\nPearson Correlation Matrix:")
display(correlation_matrix.round(4))

correlation_matrix.to_csv(
    os.path.join(
        STAT_PATH,
        "pearson_correlation_matrix.csv"
    )
)


# ------------------------------------------------------------
# 8. PEARSON CORRELATION HEATMAP
# ------------------------------------------------------------

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title(
    "Pearson Correlation Matrix"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        STAT_PATH,
        "pearson_correlation_heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 9. SPEARMAN CORRELATION ANALYSIS
# ------------------------------------------------------------

spearman_results = []

target_variable = "Cases"

predictor_variables = [
    "Temp",
    "preci",
    "LAI",
    "Deaths",
    "week_of_outbreak"
]

predictor_variables = [
    col for col in predictor_variables
    if col in df.columns
]

for variable in predictor_variables:

    temp_data = df[
        [target_variable, variable]
    ].dropna()

    if len(temp_data) < 3:
        continue

    correlation, p_value = spearmanr(
        temp_data[target_variable],
        temp_data[variable]
    )

    spearman_results.append({
        "Variable": variable,
        "Sample_Size": len(temp_data),
        "Spearman_Correlation": correlation,
        "P_Value": p_value
    })


spearman_results_df = pd.DataFrame(
    spearman_results
)

if not spearman_results_df.empty:

    spearman_results_df["Significant_at_0.05"] = (
        spearman_results_df["P_Value"] < 0.05
    )

    spearman_results_df[
        "Correlation_Strength"
    ] = spearman_results_df[
        "Spearman_Correlation"
    ].abs().apply(
        lambda x:
        "Very Weak" if x < 0.20 else
        "Weak" if x < 0.40 else
        "Moderate" if x < 0.60 else
        "Strong" if x < 0.80 else
        "Very Strong"
    )

    spearman_results_df = (
        spearman_results_df.sort_values(
            by="Spearman_Correlation",
            ascending=False
        )
    )

print("\nSpearman Correlation Results:")
display(
    spearman_results_df.round(4)
)

spearman_results_df.to_csv(
    os.path.join(
        STAT_PATH,
        "spearman_correlation_results.csv"
    ),
    index=False
)


# ------------------------------------------------------------
# 10. SPEARMAN CORRELATION BAR CHART
# ------------------------------------------------------------

if not spearman_results_df.empty:

    plt.figure(figsize=(10, 6))

    sns.barplot(
        data=spearman_results_df,
        x="Spearman_Correlation",
        y="Variable"
    )

    plt.axvline(
        0,
        linewidth=1
    )

    plt.title(
        "Spearman Correlation with Disease Cases"
    )

    plt.xlabel(
        "Spearman Correlation"
    )

    plt.ylabel(
        "Variable"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            STAT_PATH,
            "spearman_correlation.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ------------------------------------------------------------
# 11. SEASONAL CASE DISTRIBUTION
# ------------------------------------------------------------

if "Season" in df.columns:

    season_order = [
        "Summer",
        "Monsoon",
        "Post-Monsoon",
        "Winter"
    ]

    available_seasons = [
        season for season in season_order
        if season in df["Season"].dropna().unique()
    ]

    seasonal_groups = []

    for season in available_seasons:

        values = df.loc[
            df["Season"] == season,
            "Cases"
        ].dropna()

        if len(values) > 0:
            seasonal_groups.append(values)

    plt.figure(figsize=(10, 6))

    sns.boxplot(
        data=df,
        x="Season",
        y="Cases",
        order=available_seasons
    )

    plt.title(
        "Disease Cases Across Seasons"
    )

    plt.xlabel("Season")
    plt.ylabel("Cases")

    plt.xticks(rotation=20)

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            STAT_PATH,
            "seasonal_cases_boxplot.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ------------------------------------------------------------
# 12. ONE-WAY ANOVA: CASES ACROSS SEASONS
# ------------------------------------------------------------

anova_results = []

if "Season" in df.columns:

    season_groups_dict = {}

    for season in available_seasons:

        values = df.loc[
            df["Season"] == season,
            "Cases"
        ].dropna()

        if len(values) >= 2:
            season_groups_dict[season] = values

    if len(season_groups_dict) >= 2:

        groups = list(
            season_groups_dict.values()
        )

        f_statistic, p_value = f_oneway(
            *groups
        )

        anova_results.append({
            "Analysis": "Overall Seasonal ANOVA",
            "F_Statistic": f_statistic,
            "P_Value": p_value,
            "Significant_at_0.05": p_value < 0.05,
            "Number_of_Seasons": len(groups)
        })


# ------------------------------------------------------------
# 13. DISEASE-WISE SEASONAL ANOVA
# ------------------------------------------------------------

if (
    "Disease" in df.columns
    and "Season" in df.columns
):

    diseases = (
        df["Disease"]
        .dropna()
        .unique()
    )

    for disease in sorted(diseases):

        disease_data = df[
            df["Disease"] == disease
        ]

        disease_groups = []

        valid_seasons = []

        for season in available_seasons:

            values = disease_data.loc[
                disease_data["Season"] == season,
                "Cases"
            ].dropna()

            if len(values) >= 2:

                disease_groups.append(
                    values
                )

                valid_seasons.append(
                    season
                )

        if len(disease_groups) >= 2:

            try:

                f_statistic, p_value = f_oneway(
                    *disease_groups
                )

                anova_results.append({
                    "Analysis": "Disease-wise Seasonal ANOVA",
                    "Disease": disease,
                    "F_Statistic": f_statistic,
                    "P_Value": p_value,
                    "Significant_at_0.05": (
                        p_value < 0.05
                    ),
                    "Number_of_Seasons": (
                        len(disease_groups)
                    )
                })

            except Exception as e:

                print(
                    f"ANOVA failed for {disease}: {e}"
                )


# ------------------------------------------------------------
# 14. SAVE ANOVA RESULTS
# ------------------------------------------------------------

anova_results_df = pd.DataFrame(
    anova_results
)

print("\nANOVA Results:")

if not anova_results_df.empty:
    display(
        anova_results_df.round(4)
    )

    anova_results_df.to_csv(
        os.path.join(
            STAT_PATH,
            "seasonal_anova_results.csv"
        ),
        index=False
    )

else:

    print(
        "No valid ANOVA results were generated."
    )


# ------------------------------------------------------------
# 15. SEASONAL DESCRIPTIVE STATISTICS
# ------------------------------------------------------------

if "Season" in df.columns:

    seasonal_statistics = (
        df.groupby("Season")["Cases"]
        .agg([
            "count",
            "mean",
            "median",
            "std",
            "min",
            "max",
            "sum"
        ])
        .reindex(available_seasons)
    )

    seasonal_statistics[
        "coefficient_of_variation"
    ] = (
        seasonal_statistics["std"]
        / seasonal_statistics["mean"]
    )

    seasonal_statistics = (
        seasonal_statistics.round(4)
    )

    print("\nSeasonal Descriptive Statistics:")
    display(seasonal_statistics)

    seasonal_statistics.to_csv(
        os.path.join(
            STAT_PATH,
            "seasonal_descriptive_statistics.csv"
        )
    )


# ------------------------------------------------------------
# 16. DISEASE-WISE SEASONAL MEAN CASES
# ------------------------------------------------------------

if (
    "Disease" in df.columns
    and "Season" in df.columns
):

    disease_season_mean = (
        df.groupby(
            ["Disease", "Season"]
        )["Cases"]
        .agg([
            "count",
            "mean",
            "median",
            "sum"
        ])
        .reset_index()
    )

    disease_season_mean = (
        disease_season_mean.round(4)
    )

    print(
        "\nDisease-wise Seasonal Statistics:"
    )

    display(
        disease_season_mean.head(20)
    )

    disease_season_mean.to_csv(
        os.path.join(
            STAT_PATH,
            "disease_season_statistics.csv"
        ),
        index=False
    )


# ------------------------------------------------------------
# 17. DISEASE × SEASON HEATMAP
# ------------------------------------------------------------

if (
    "Disease" in df.columns
    and "Season" in df.columns
):

    disease_season_pivot = pd.pivot_table(
        df,
        values="Cases",
        index="Disease",
        columns="Season",
        aggfunc="sum",
        fill_value=0
    )

    disease_season_pivot = (
        disease_season_pivot.reindex(
            columns=available_seasons
        )
    )

    print(
        "\nDisease × Season Case Distribution:"
    )

    display(
        disease_season_pivot
    )

    disease_season_pivot.to_csv(
        os.path.join(
            STAT_PATH,
            "disease_season_case_pivot.csv"
        )
    )

    plt.figure(
        figsize=(12, 8)
    )

    sns.heatmap(
        disease_season_pivot,
        annot=True,
        fmt=".0f",
        cmap="YlOrRd"
    )

    plt.title(
        "Disease Cases by Season"
    )

    plt.xlabel("Season")
    plt.ylabel("Disease")

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            STAT_PATH,
            "disease_season_heatmap.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ------------------------------------------------------------
# 18. TEMPERATURE vs CASES
# ------------------------------------------------------------

if (
    "Temp" in df.columns
    and "Cases" in df.columns
):

    temp_cases = df[
        ["Temp", "Cases"]
    ].dropna()

    if len(temp_cases) > 0:

        plt.figure(figsize=(9, 6))

        sns.scatterplot(
            data=temp_cases,
            x="Temp",
            y="Cases",
            alpha=0.5
        )

        plt.title(
            "Temperature vs Disease Cases"
        )

        plt.xlabel(
            "Temperature"
        )

        plt.ylabel(
            "Disease Cases"
        )

        plt.tight_layout()

        plt.savefig(
            os.path.join(
                STAT_PATH,
                "temperature_vs_cases.png"
            ),
            dpi=300,
            bbox_inches="tight"
        )

        plt.show()


# ------------------------------------------------------------
# 19. PRECIPITATION vs CASES
# ------------------------------------------------------------

if (
    "preci" in df.columns
    and "Cases" in df.columns
):

    precipitation_cases = df[
        ["preci", "Cases"]
    ].dropna()

    if len(precipitation_cases) > 0:

        plt.figure(figsize=(9, 6))

        sns.scatterplot(
            data=precipitation_cases,
            x="preci",
            y="Cases",
            alpha=0.5
        )

        plt.title(
            "Precipitation vs Disease Cases"
        )

        plt.xlabel(
            "Precipitation"
        )

        plt.ylabel(
            "Disease Cases"
        )

        plt.tight_layout()

        plt.savefig(
            os.path.join(
                STAT_PATH,
                "precipitation_vs_cases.png"
            ),
            dpi=300,
            bbox_inches="tight"
        )

        plt.show()


# ------------------------------------------------------------
# 20. LAI vs CASES
# ------------------------------------------------------------

if (
    "LAI" in df.columns
    and "Cases" in df.columns
):

    lai_cases = df[
        ["LAI", "Cases"]
    ].dropna()

    if len(lai_cases) > 0:

        plt.figure(figsize=(9, 6))

        sns.scatterplot(
            data=lai_cases,
            x="LAI",
            y="Cases",
            alpha=0.5
        )

        plt.title(
            "Leaf Area Index (LAI) vs Disease Cases"
        )

        plt.xlabel(
            "LAI"
        )

        plt.ylabel(
            "Disease Cases"
        )

        plt.tight_layout()

        plt.savefig(
            os.path.join(
                STAT_PATH,
                "lai_vs_cases.png"
            ),
            dpi=300,
            bbox_inches="tight"
        )

        plt.show()


# ------------------------------------------------------------
# 21. YEARLY CASE TREND
# ------------------------------------------------------------

if (
    "Year" in df.columns
    and "Cases" in df.columns
):

    yearly_cases = (
        df.groupby("Year")["Cases"]
        .sum()
        .reset_index()
    )

    print("\nYearly Case Trend:")
    display(yearly_cases)

    yearly_cases.to_csv(
        os.path.join(
            STAT_PATH,
            "yearly_case_trend.csv"
        ),
        index=False
    )

    plt.figure(figsize=(12, 6))

    sns.lineplot(
        data=yearly_cases,
        x="Year",
        y="Cases",
        marker="o"
    )

    plt.title(
        "Yearly Disease Case Trend"
    )

    plt.xlabel("Year")
    plt.ylabel("Total Cases")

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            STAT_PATH,
            "yearly_case_trend.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ------------------------------------------------------------
# 22. GENERATE FINAL STATISTICAL SUMMARY
# ------------------------------------------------------------

summary_rows = []

summary_rows.append({
    "Metric": "Total Records",
    "Value": len(df)
})

summary_rows.append({
    "Metric": "Total Diseases",
    "Value": df["Disease"].nunique()
    if "Disease" in df.columns else np.nan
})

summary_rows.append({
    "Metric": "Total Districts",
    "Value": df["district"].nunique()
    if "district" in df.columns else np.nan
})

summary_rows.append({
    "Metric": "Total Cases",
    "Value": df["Cases"].sum()
    if "Cases" in df.columns else np.nan
})

if "Season" in df.columns:

    summary_rows.append({
        "Metric": "Most Common Season",
        "Value": df["Season"].mode()[0]
        if not df["Season"].mode().empty
        else np.nan
    })

if (
    not spearman_results_df.empty
    and "P_Value" in spearman_results_df.columns
):

    significant_variables = (
        spearman_results_df.loc[
            spearman_results_df[
                "P_Value"
            ] < 0.05,
            "Variable"
        ].tolist()
    )

    summary_rows.append({
        "Metric": "Significant Spearman Variables (p < 0.05)",
        "Value": ", ".join(
            significant_variables
        )
        if significant_variables
        else "None"
    })


if (
    not anova_results_df.empty
    and "P_Value" in anova_results_df.columns
):

    significant_anova_count = (
        anova_results_df[
            "P_Value"
        ] < 0.05
    ).sum()

    summary_rows.append({
        "Metric": "Significant ANOVA Tests (p < 0.05)",
        "Value": significant_anova_count
    })


final_summary = pd.DataFrame(
    summary_rows
)

print("\nFINAL STATISTICAL ANALYSIS SUMMARY")
display(final_summary)

final_summary.to_csv(
    os.path.join(
        STAT_PATH,
        "final_statistical_summary.csv"
    ),
    index=False
)


# ------------------------------------------------------------
# 23. LIST GENERATED FILES
# ------------------------------------------------------------

print("\n============================================================")
print("STATISTICAL ANALYSIS COMPLETED")
print("============================================================")

print("\nGenerated files:")

for file_name in sorted(
    os.listdir(STAT_PATH)
):

    print("✓", file_name)

print("\nOutput folder:")
print(STAT_PATH)
```
