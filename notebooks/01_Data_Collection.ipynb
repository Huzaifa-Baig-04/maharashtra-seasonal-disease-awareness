```python
# ============================================================
# CEP SEASONAL DISEASE PREDICTION
# FILE 01: DATA COLLECTION
# ============================================================

# ------------------------------------------------------------
# 1. Mount Google Drive
# ------------------------------------------------------------

from google.colab import drive
drive.mount("/content/drive")


# ------------------------------------------------------------
# 2. Import Libraries
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import os


# ------------------------------------------------------------
# 3. Define Project Paths
# ------------------------------------------------------------

PROJECT_PATH = "/content/drive/MyDrive/CEP Seasonal Disease Prediction"

RAW_PATH = os.path.join(
    PROJECT_PATH,
    "Data",
    "raw"
)

PROCESSED_PATH = os.path.join(
    PROJECT_PATH,
    "Data",
    "processed"
)

# Create folders if they don't exist
os.makedirs(RAW_PATH, exist_ok=True)
os.makedirs(PROCESSED_PATH, exist_ok=True)


# ------------------------------------------------------------
# 4. Dataset Path
# ------------------------------------------------------------

DATASET_PATH = os.path.join(
    RAW_PATH,
    "Final_data.csv"
)

print("=" * 60)
print("CEP SEASONAL DISEASE PREDICTION")
print("01 - DATA COLLECTION")
print("=" * 60)

print("\nProject Path:")
print(PROJECT_PATH)

print("\nDataset Path:")
print(DATASET_PATH)


# ------------------------------------------------------------
# 5. Check Dataset Exists
# ------------------------------------------------------------

if not os.path.exists(DATASET_PATH):
    raise FileNotFoundError(
        f"Final_data.csv was not found at:\n{DATASET_PATH}\n\n"
        "Please place the raw dataset inside:\n"
        "Data/raw/"
    )

print("\n✓ Final_data.csv found successfully.")


# ------------------------------------------------------------
# 6. Load Raw Dataset
# ------------------------------------------------------------

df = pd.read_csv(DATASET_PATH)

print("\nDataset loaded successfully.")


# ------------------------------------------------------------
# 7. Basic Dataset Information
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

print("\nRows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)


# ------------------------------------------------------------
# 8. First Five Records
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FIRST 5 RECORDS")
print("=" * 60)

display(df.head())


# ------------------------------------------------------------
# 9. Dataset Information
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("DATAFRAME INFO")
print("=" * 60)

df.info()


# ------------------------------------------------------------
# 10. Statistical Summary
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("STATISTICAL SUMMARY")
print("=" * 60)

display(df.describe(include="all").T)


# ------------------------------------------------------------
# 11. Missing Values
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MISSING VALUES")
print("=" * 60)

missing_values = df.isnull().sum()

missing_percentage = (
    missing_values / len(df) * 100
)

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": missing_percentage
})

display(
    missing_summary.sort_values(
        by="Missing Values",
        ascending=False
    )
)


# ------------------------------------------------------------
# 12. Duplicate Records
# ------------------------------------------------------------

duplicate_count = df.duplicated().sum()

print("\nDuplicate Records:", duplicate_count)


# ------------------------------------------------------------
# 13. States / UTs Available
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("STATE / UT DISTRIBUTION")
print("=" * 60)

display(
    df["state_ut"].value_counts()
)


# ------------------------------------------------------------
# 14. Maharashtra Records
# ------------------------------------------------------------

maharashtra_df = df[
    df["state_ut"]
    .astype(str)
    .str.strip()
    .str.lower()
    .eq("maharashtra")
].copy()

print("\n" + "=" * 60)
print("MAHARASHTRA DATA")
print("=" * 60)

print(
    "Maharashtra Records:",
    len(maharashtra_df)
)


# ------------------------------------------------------------
# 15. Maharashtra Districts
# ------------------------------------------------------------

print(
    "\nNumber of Maharashtra district labels:",
    maharashtra_df["district"].nunique()
)

print("\nDistricts:")

display(
    pd.DataFrame({
        "District": sorted(
            maharashtra_df["district"]
            .dropna()
            .unique()
        )
    })
)


# ------------------------------------------------------------
# 16. Maharashtra Disease Categories
# ------------------------------------------------------------

print(
    "\nNumber of Disease Categories:",
    maharashtra_df["Disease"].nunique()
)

print("\nDisease Distribution:")

display(
    maharashtra_df["Disease"]
    .value_counts()
)


# ------------------------------------------------------------
# 17. Maharashtra Year Range
# ------------------------------------------------------------

print(
    "\nMinimum Year:",
    maharashtra_df["year"].min()
)

print(
    "Maximum Year:",
    maharashtra_df["year"].max()
)


# ------------------------------------------------------------
# 18. Maharashtra Date Range
# ------------------------------------------------------------

if all(
    column in maharashtra_df.columns
    for column in ["year", "mon", "day"]
):

    maharashtra_df["Date"] = pd.to_datetime(
        dict(
            year=maharashtra_df["year"],
            month=maharashtra_df["mon"],
            day=maharashtra_df["day"]
        ),
        errors="coerce"
    )

    print(
        "\nMaharashtra Date Range:"
    )

    print(
        maharashtra_df["Date"].min(),
        "to",
        maharashtra_df["Date"].max()
    )


# ------------------------------------------------------------
# 19. Dataset Source Information
# ------------------------------------------------------------

dataset_information = {
    "Dataset_Name":
        "EpiClim Weekly District-Wise All-India Climate-Health Dataset",

    "Raw_File":
        "Final_data.csv",

    "Project_State":
        "Maharashtra",

    "Analysis_Level":
        "District",

    "Data_Type":
        "Disease and climate-health data",

    "Project_Title":
        "Predictive Analysis of Seasonal Diseases in Maharashtra Using Data Science"
}


# ------------------------------------------------------------
# 20. Save Dataset Information
# ------------------------------------------------------------

information_file = os.path.join(
    RAW_PATH,
    "dataset_information.csv"
)

pd.DataFrame(
    [dataset_information]
).to_csv(
    information_file,
    index=False
)

print(
    "\n✓ Dataset information saved to:"
)

print(information_file)


# ------------------------------------------------------------
# 21. Final Data Collection Summary
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("DATA COLLECTION SUMMARY")
print("=" * 60)

print(
    f"\nOriginal Dataset      : {df.shape[0]} rows × {df.shape[1]} columns"
)

print(
    f"Maharashtra Records  : {len(maharashtra_df)}"
)

print(
    f"Maharashtra Districts: {maharashtra_df['district'].nunique()}"
)

print(
    f"Disease Categories   : {maharashtra_df['Disease'].nunique()}"
)

print(
    f"Year Range            : "
    f"{maharashtra_df['year'].min()} - "
    f"{maharashtra_df['year'].max()}"
)

print(
    f"Duplicate Records     : {duplicate_count}"
)

print(
    f"Missing Values        : {df.isnull().sum().sum()}"
)

print("\n✓ DATA COLLECTION COMPLETED")
print("=" * 60)

print(
    "\nIMPORTANT:"
    "\nThe original Final_data.csv has NOT been modified."
    "\nCleaning and Maharashtra-specific processing will be done"
    "\nin 02_Data_Cleaning.ipynb."
)
```
