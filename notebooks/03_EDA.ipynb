```python
# ============================================================
# CEP SEASONAL DISEASE PREDICTION
# FILE 03: EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================

# ------------------------------------------------------------
# 1. Mount Google Drive
# ------------------------------------------------------------

from google.colab import drive
drive.mount("/content/drive")


# ------------------------------------------------------------
# 2. Import Libraries
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os


# ------------------------------------------------------------
# 3. Define Project Paths
# ------------------------------------------------------------

PROJECT_PATH = "/content/drive/MyDrive/CEP Seasonal Disease Prediction"

PROCESSED_PATH = os.path.join(
    PROJECT_PATH,
    "Data",
    "processed"
)

EDA_PATH = os.path.join(
    PROCESSED_PATH,
    "EDA"
)

os.makedirs(EDA_PATH, exist_ok=True)


# ------------------------------------------------------------
# 4. Load Cleaned Maharashtra Dataset
# ------------------------------------------------------------

INPUT_FILE = os.path.join(
    PROCESSED_PATH,
    "Maharashtra_Seasonal_Disease_Cleaned.csv"
)

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"Cleaned dataset not found:\n{INPUT_FILE}\n\n"
        "Please run 02_Data_Cleaning.ipynb first."
    )

df = pd.read_csv(INPUT_FILE)

print("=" * 70)
print("CEP SEASONAL DISEASE PREDICTION")
print("03 - EXPLORATORY DATA ANALYSIS")
print("=" * 70)

print("\nDataset Shape:")
print(df.shape)


# ------------------------------------------------------------
# 5. Convert Date Column
# ------------------------------------------------------------

df["Date"] = pd.to_datetime(
    df["Date"],
    errors="coerce"
)


# ------------------------------------------------------------
# 6. Basic Dataset Information
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("BASIC DATASET INFORMATION")
print("=" * 70)

print("\nRows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)


# ------------------------------------------------------------
# 7. Statistical Summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STATISTICAL SUMMARY")
print("=" * 70)

display(
    df.describe(include="all").T
)


# ------------------------------------------------------------
# 8. Missing Value Analysis
# ------------------------------------------------------------

missing_values = df.isnull().sum()

missing_percentage = (
    missing_values / len(df) * 100
)

missing_summary = pd.DataFrame({
    "Missing_Values": missing_values,
    "Percentage": missing_percentage
})

missing_summary = missing_summary.sort_values(
    by="Missing_Values",
    ascending=False
)

print("\n" + "=" * 70)
print("MISSING VALUE ANALYSIS")
print("=" * 70)

display(missing_summary)


# ------------------------------------------------------------
# 9. Missing Values Plot
# ------------------------------------------------------------

missing_plot_data = missing_summary[
    missing_summary["Missing_Values"] > 0
]

if len(missing_plot_data) > 0:

    plt.figure(figsize=(12, 6))

    plt.bar(
        missing_plot_data.index,
        missing_plot_data["Missing_Values"]
    )

    plt.title(
        "Missing Values by Column"
    )

    plt.xlabel("Column")
    plt.ylabel("Number of Missing Values")

    plt.xticks(
        rotation=45,
        ha="right"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            EDA_PATH,
            "01_missing_values.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ------------------------------------------------------------
# 10. Duplicate Analysis
# ------------------------------------------------------------

duplicate_count = df.duplicated().sum()

print("\n" + "=" * 70)
print("DUPLICATE ANALYSIS")
print("=" * 70)

print(
    "Duplicate records:",
    duplicate_count
)


# ------------------------------------------------------------
# 11. District Analysis
# ------------------------------------------------------------

district_counts = (
    df["district"]
    .value_counts()
)

print("\n" + "=" * 70)
print("DISTRICT ANALYSIS")
print("=" * 70)

print(
    "Number of districts:",
    df["district"].nunique()
)

display(district_counts)


# ------------------------------------------------------------
# 12. District Record Distribution
# ------------------------------------------------------------

plt.figure(figsize=(14, 8))

district_counts.sort_values(
    ascending=True
).plot(
    kind="barh"
)

plt.title(
    "Disease Records by Maharashtra District"
)

plt.xlabel("Number of Records")
plt.ylabel("District")

plt.tight_layout()

plt.savefig(
    os.path.join(
        EDA_PATH,
        "02_district_record_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 13. Disease Analysis
# ------------------------------------------------------------

disease_counts = (
    df["Disease"]
    .value_counts()
)

print("\n" + "=" * 70)
print("DISEASE ANALYSIS")
print("=" * 70)

print(
    "Number of disease categories:",
    df["Disease"].nunique()
)

display(disease_counts)


# ------------------------------------------------------------
# 14. Disease Distribution Plot
# ------------------------------------------------------------

plt.figure(figsize=(12, 8))

disease_counts.sort_values(
    ascending=True
).plot(
    kind="barh"
)

plt.title(
    "Disease Distribution in Maharashtra"
)

plt.xlabel("Number of Records")
plt.ylabel("Disease")

plt.tight_layout()

plt.savefig(
    os.path.join(
        EDA_PATH,
        "03_disease_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 15. Total Cases by Disease
# ------------------------------------------------------------

cases_by_disease = (
    df.groupby("Disease")["Cases"]
    .sum()
    .sort_values(
        ascending=False
    )
)

print("\n" + "=" * 70)
print("TOTAL CASES BY DISEASE")
print("=" * 70)

display(cases_by_disease)


# ------------------------------------------------------------
# 16. Total Cases by Disease Plot
# ------------------------------------------------------------

plt.figure(figsize=(12, 8))

cases_by_disease.sort_values(
    ascending=True
).plot(
    kind="barh"
)

plt.title(
    "Total Reported Cases by Disease"
)

plt.xlabel("Total Cases")
plt.ylabel("Disease")

plt.tight_layout()

plt.savefig(
    os.path.join(
        EDA_PATH,
        "04_total_cases_by_disease.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 17. Yearly Disease Cases
# ------------------------------------------------------------

yearly_cases = (
    df.groupby("year")["Cases"]
    .sum()
    .sort_index()
)

print("\n" + "=" * 70)
print("YEARLY DISEASE CASES")
print("=" * 70)

display(yearly_cases)


# ------------------------------------------------------------
# 18. Yearly Cases Plot
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

plt.plot(
    yearly_cases.index,
    yearly_cases.values,
    marker="o"
)

plt.title(
    "Total Disease Cases by Year in Maharashtra"
)

plt.xlabel("Year")
plt.ylabel("Total Cases")

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        EDA_PATH,
        "05_yearly_cases.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 19. Monthly Disease Cases
# ------------------------------------------------------------

monthly_cases = (
    df.groupby("mon")["Cases"]
    .sum()
    .sort_index()
)

print("\n" + "=" * 70)
print("MONTHLY DISEASE CASES")
print("=" * 70)

display(monthly_cases)


# ------------------------------------------------------------
# 20. Monthly Cases Plot
# ------------------------------------------------------------

month_names = {
    1: "Jan",
    2: "Feb",
    3: "Mar",
    4: "Apr",
    5: "May",
    6: "Jun",
    7: "Jul",
    8: "Aug",
    9: "Sep",
    10: "Oct",
    11: "Nov",
    12: "Dec"
}

monthly_plot = monthly_cases.copy()

monthly_plot.index = [
    month_names.get(
        month,
        str(month)
    )
    for month in monthly_plot.index
]

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_plot.index,
    monthly_plot.values,
    marker="o"
)

plt.title(
    "Total Disease Cases by Month"
)

plt.xlabel("Month")
plt.ylabel("Total Cases")

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        EDA_PATH,
        "06_monthly_cases.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 21. Seasonal Analysis
# ------------------------------------------------------------

seasonal_cases = (
    df.groupby("Season")["Cases"]
    .sum()
    .sort_values(
        ascending=False
    )
)

print("\n" + "=" * 70)
print("SEASONAL DISEASE CASES")
print("=" * 70)

display(seasonal_cases)


# ------------------------------------------------------------
# 22. Seasonal Cases Plot
# ------------------------------------------------------------

season_order = [
    "Winter",
    "Summer",
    "Monsoon",
    "Post-Monsoon"
]

seasonal_plot = (
    df.groupby("Season")["Cases"]
    .sum()
    .reindex(season_order)
)

plt.figure(figsize=(10, 6))

seasonal_plot.plot(
    kind="bar"
)

plt.title(
    "Total Disease Cases by Season"
)

plt.xlabel("Season")
plt.ylabel("Total Cases")

plt.xticks(
    rotation=0
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        EDA_PATH,
        "07_seasonal_cases.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 23. Disease × Season Analysis
# ------------------------------------------------------------

disease_season = pd.pivot_table(
    df,
    values="Cases",
    index="Disease",
    columns="Season",
    aggfunc="sum",
    fill_value=0
)

disease_season = disease_season.reindex(
    columns=[
        column
        for column in season_order
        if column in disease_season.columns
    ]
)

print("\n" + "=" * 70)
print("DISEASE × SEASON")
print("=" * 70)

display(disease_season)


# ------------------------------------------------------------
# 24. Disease × Season Heatmap
# ------------------------------------------------------------

plt.figure(figsize=(14, 8))

plt.imshow(
    disease_season,
    aspect="auto"
)

plt.colorbar(
    label="Total Cases"
)

plt.xticks(
    range(len(disease_season.columns)),
    disease_season.columns,
    rotation=45
)

plt.yticks(
    range(len(disease_season.index)),
    disease_season.index
)

plt.title(
    "Disease Cases Across Seasons"
)

plt.xlabel("Season")
plt.ylabel("Disease")

plt.tight_layout()

plt.savefig(
    os.path.join(
        EDA_PATH,
        "08_disease_season_heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 25. Disease × Month Analysis
# ------------------------------------------------------------

disease_month = pd.pivot_table(
    df,
    values="Cases",
    index="Disease",
    columns="mon",
    aggfunc="sum",
    fill_value=0
)

disease_month = disease_month.reindex(
    columns=range(1, 13),
    fill_value=0
)

print("\n" + "=" * 70)
print("DISEASE × MONTH")
print("=" * 70)

display(disease_month)


# ------------------------------------------------------------
# 26. Disease × Month Heatmap
# ------------------------------------------------------------

plt.figure(figsize=(15, 8))

plt.imshow(
    disease_month,
    aspect="auto"
)

plt.colorbar(
    label="Total Cases"
)

plt.xticks(
    range(12),
    [
        "Jan", "Feb", "Mar", "Apr",
        "May", "Jun", "Jul", "Aug",
        "Sep", "Oct", "Nov", "Dec"
    ],
    rotation=45
)

plt.yticks(
    range(len(disease_month.index)),
    disease_month.index
)

plt.title(
    "Disease Cases by Month"
)

plt.xlabel("Month")
plt.ylabel("Disease")

plt.tight_layout()

plt.savefig(
    os.path.join(
        EDA_PATH,
        "09_disease_month_heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 27. District-wise Total Cases
# ------------------------------------------------------------

district_cases = (
    df.groupby("district")["Cases"]
    .sum()
    .sort_values(
        ascending=False
    )
)

print("\n" + "=" * 70)
print("DISTRICT-WISE TOTAL CASES")
print("=" * 70)

display(
    district_cases
)


# ------------------------------------------------------------
# 28. Top 15 Districts Plot
# ------------------------------------------------------------

top_15_districts = (
    district_cases
    .head(15)
    .sort_values()
)

plt.figure(figsize=(12, 8))

top_15_districts.plot(
    kind="barh"
)

plt.title(
    "Top 15 Maharashtra Districts by Total Disease Cases"
)

plt.xlabel("Total Cases")
plt.ylabel("District")

plt.tight_layout()

plt.savefig(
    os.path.join(
        EDA_PATH,
        "10_top_15_districts.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 29. Temperature vs Cases
# ------------------------------------------------------------

temperature_data = df[
    ["Temp", "Cases"]
].dropna()

plt.figure(figsize=(9, 6))

plt.scatter(
    temperature_data["Temp"],
    temperature_data["Cases"],
    alpha=0.5
)

plt.title(
    "Temperature vs Disease Cases"
)

plt.xlabel(
    "Temperature (Kelvin)"
)

plt.ylabel(
    "Disease Cases"
)

plt.grid(
    True,
    alpha=0.2
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        EDA_PATH,
        "11_temperature_vs_cases.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 30. Temperature in Celsius
# ------------------------------------------------------------

df["Temperature_C"] = (
    df["Temp"] - 273.15
)

temperature_c_data = df[
    ["Temperature_C", "Cases"]
].dropna()

plt.figure(figsize=(9, 6))

plt.scatter(
    temperature_c_data["Temperature_C"],
    temperature_c_data["Cases"],
    alpha=0.5
)

plt.title(
    "Temperature vs Disease Cases"
)

plt.xlabel(
    "Temperature (°C)"
)

plt.ylabel(
    "Disease Cases"
)

plt.grid(
    True,
    alpha=0.2
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        EDA_PATH,
        "12_temperature_celsius_vs_cases.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 31. Precipitation vs Cases
# ------------------------------------------------------------

precipitation_data = df[
    ["preci", "Cases"]
].dropna()

plt.figure(figsize=(9, 6))

plt.scatter(
    precipitation_data["preci"],
    precipitation_data["Cases"],
    alpha=0.5
)

plt.title(
    "Precipitation vs Disease Cases"
)

plt.xlabel(
    "Precipitation"
)

plt.ylabel(
    "Disease Cases"
)

plt.grid(
    True,
    alpha=0.2
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        EDA_PATH,
        "13_precipitation_vs_cases.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 32. LAI vs Cases
# ------------------------------------------------------------

lai_data = df[
    ["LAI", "Cases"]
].dropna()

plt.figure(figsize=(9, 6))

plt.scatter(
    lai_data["LAI"],
    lai_data["Cases"],
    alpha=0.5
)

plt.title(
    "Leaf Area Index (LAI) vs Disease Cases"
)

plt.xlabel(
    "LAI"
)

plt.ylabel(
    "Disease Cases"
)

plt.grid(
    True,
    alpha=0.2
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        EDA_PATH,
        "14_lai_vs_cases.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 33. Numerical Correlation Analysis
# ------------------------------------------------------------

numeric_columns = [
    "Cases",
    "Deaths",
    "preci",
    "LAI",
    "Temp",
    "week_of_outbreak"
]

available_numeric_columns = [
    column
    for column in numeric_columns
    if column in df.columns
]

correlation = (
    df[available_numeric_columns]
    .corr()
)

print("\n" + "=" * 70)
print("CORRELATION MATRIX")
print("=" * 70)

display(correlation)


# ------------------------------------------------------------
# 34. Correlation Heatmap
# ------------------------------------------------------------

plt.figure(figsize=(10, 8))

plt.imshow(
    correlation,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.colorbar(
    label="Correlation"
)

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation.index)),
    correlation.index
)

plt.title(
    "Correlation Matrix of Disease and Environmental Variables"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        EDA_PATH,
        "15_correlation_heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 35. Peak Month for Each Disease
# ------------------------------------------------------------

peak_month = (
    df.groupby(
        ["Disease", "mon"]
    )["Cases"]
    .sum()
    .reset_index()
)

peak_month = peak_month.loc[
    peak_month.groupby(
        "Disease"
    )["Cases"].idxmax()
]

peak_month = peak_month.sort_values(
    "Cases",
    ascending=False
)

print("\n" + "=" * 70)
print("PEAK MONTH FOR EACH DISEASE")
print("=" * 70)

display(peak_month)


# ------------------------------------------------------------
# 36. Add Month Names
# ------------------------------------------------------------

peak_month["Month_Name"] = (
    peak_month["mon"]
    .map(month_names)
)


# ------------------------------------------------------------
# 37. Peak Season for Each Disease
# ------------------------------------------------------------

peak_season = (
    df.groupby(
        ["Disease", "Season"]
    )["Cases"]
    .sum()
    .reset_index()
)

peak_season = peak_season.loc[
    peak_season.groupby(
        "Disease"
    )["Cases"].idxmax()
]

peak_season = peak_season.sort_values(
    "Cases",
    ascending=False
)

print("\n" + "=" * 70)
print("PEAK SEASON FOR EACH DISEASE")
print("=" * 70)

display(peak_season)


# ------------------------------------------------------------
# 38. Save EDA Tables
# ------------------------------------------------------------

disease_counts.to_csv(
    os.path.join(
        EDA_PATH,
        "disease_counts.csv"
    ),
    header=["Record_Count"]
)

cases_by_disease.to_csv(
    os.path.join(
        EDA_PATH,
        "cases_by_disease.csv"
    ),
    header=["Total_Cases"]
)

yearly_cases.to_csv(
    os.path.join(
        EDA_PATH,
        "yearly_cases.csv"
    ),
    header=["Total_Cases"]
)

monthly_cases.to_csv(
    os.path.join(
        EDA_PATH,
        "monthly_cases.csv"
    ),
    header=["Total_Cases"]
)

seasonal_cases.to_csv(
    os.path.join(
        EDA_PATH,
        "seasonal_cases.csv"
    ),
    header=["Total_Cases"]
)

disease_season.to_csv(
    os.path.join(
        EDA_PATH,
        "disease_season.csv"
    )
)

disease_month.to_csv(
    os.path.join(
        EDA_PATH,
        "disease_month.csv"
    )
)

district_cases.to_csv(
    os.path.join(
        EDA_PATH,
        "district_cases.csv"
    ),
    header=["Total_Cases"]
)

peak_month.to_csv(
    os.path.join(
        EDA_PATH,
        "peak_month_by_disease.csv"
    ),
    index=False
)

peak_season.to_csv(
    os.path.join(
        EDA_PATH,
        "peak_season_by_disease.csv"
    ),
    index=False
)

missing_summary.to_csv(
    os.path.join(
        EDA_PATH,
        "missing_value_summary.csv"
    )
)

correlation.to_csv(
    os.path.join(
        EDA_PATH,
        "correlation_matrix.csv"
    )
)


# ------------------------------------------------------------
# 39. Create Overall EDA Summary
# ------------------------------------------------------------

eda_summary = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Total Districts",
        "Total Disease Categories",
        "Minimum Year",
        "Maximum Year",
        "Total Reported Cases",
        "Total Reported Deaths",
        "Duplicate Records",
        "Missing Cases",
        "Missing Deaths",
        "Missing Temperature",
        "Missing Precipitation",
        "Missing LAI"
    ],

    "Value": [
        len(df),
        df["district"].nunique(),
        df["Disease"].nunique(),
        df["year"].min(),
        df["year"].max(),
        df["Cases"].sum(),
        df["Deaths"].sum(),
        duplicate_count,
        df["Cases"].isna().sum(),
        df["Deaths"].isna().sum(),
        df["Temp"].isna().sum(),
        df["preci"].isna().sum(),
        df["LAI"].isna().sum()
    ]
})

print("\n" + "=" * 70)
print("EDA SUMMARY")
print("=" * 70)

display(eda_summary)


# ------------------------------------------------------------
# 40. Save EDA Summary
# ------------------------------------------------------------

eda_summary.to_csv(
    os.path.join(
        EDA_PATH,
        "EDA_summary.csv"
    ),
    index=False
)


# ------------------------------------------------------------
# 41. Final Output
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("EDA COMPLETED SUCCESSFULLY")
print("=" * 70)

print("\nEDA files saved in:")

print(EDA_PATH)

print("\nGenerated outputs include:")

for file in sorted(os.listdir(EDA_PATH)):
    print(" -", file)

print("\n" + "=" * 70)
```
